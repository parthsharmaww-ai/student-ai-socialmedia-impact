```python
from scipy.stats import shapiro
import matplotlib.pyplot as plt
```

```python
continuous_cols = ['Daily_Social_Media_Hours', 'Daily_AI_Tool_Usage_Hours', 'Sleep_Hours', 'Physical_Activity_Hours', 'Mental_Health_Score', 'Physical_Health_Score', 'Social_Isolation_Score', 'Academic_Performance_Score']

for col in continuous_cols:
    plt.hist(df[col], bins=30)
    plt.title(col)
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
    stat, p_value = shapiro(df[col])
    print(col, '- statistic:', stat, ', p-value:', p_value)
```

```python
from scipy.stats import kruskal

low = df[df['Burnout_Level'] == 'Low']['Daily_Social_Media_Hours']
moderate = df[df['Burnout_Level'] == 'Moderate']['Daily_Social_Media_Hours']
high = df[df['Burnout_Level'] == 'High']['Daily_Social_Media_Hours']
severe = df[df['Burnout_Level'] == 'Severe']['Daily_Social_Media_Hours']
stat, p_value = kruskal(low, moderate, high, severe) 
print("E7 - H-statistic:", stat, ", p-value:", p_value)
```

```python
from scipy.stats import kruskal

low = df[df['Burnout_Level'] == 'Low']['Daily_AI_Tool_Usage_Hours']
moderate = df[df['Burnout_Level'] == 'Moderate']['Daily_AI_Tool_Usage_Hours']
high = df[df['Burnout_Level'] == 'High']['Daily_AI_Tool_Usage_Hours']
severe = df[df['Burnout_Level'] == 'Severe']['Daily_AI_Tool_Usage_Hours']
stat, p_value = kruskal(low, moderate, high, severe) 
print("E8 - H-statistic:", stat, ", p-value:", p_value)
```

```python
from scipy.stats import mannwhitneyu

group0 = df[df['Academic_Failure_Risk'] == False]['Daily_Social_Media_Hours']
group1 = df[df['Academic_Failure_Risk'] == True]['Daily_Social_Media_Hours']

stat, p_value = mannwhitneyu(group0, group1)
print('E10 - U-statistic:', stat, ', p-value:', p_value)
```
