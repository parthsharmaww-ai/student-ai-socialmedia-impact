```python
import pandas as pd
```

```python
df = pd.read_csv('../data/cleaned_student_data.csv')
```

```python
df['Academic_Failure_Risk'].dtype
```

```python
from scipy.stats import shapiro
import matplotlib.pyplot as plt
```

```python
continuous_cols = ['Daily_Social_Media_Hours', 'Daily_AI_Tool_Usage_Hours', 'Sleep_Hours', 'Physical_Activity_Hours', 'Mental_Health_Score', 'Physical_Health_Score', 'Social_Isolation_Score', 'Academic_Performance_Score']

for col in continuous_cols:
    plt.hist(df[col], bins=30)
    plt.title(col)
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
    stat, p_value = shapiro(df[col])
    print(col, '- statistic:', stat, ', p-value:', p_value)
```

```python
from scipy.stats import spearmanr

corr, p_value = spearmanr(df['Daily_Social_Media_Hours'], df['Sleep_Hours'])
print('H1 - correlation:', corr, ', p-value:', p_value)
```
